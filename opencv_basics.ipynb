```python
import cv2
```

read, resize and show image

```python
img = cv2.imread('bottle.png')
resize = cv2.resize(img, (500, 500))
cv2.imshow('image', resize)
cv2.imshow('image1', img)
cv2.waitKey(10000)
cv2.destroyAllWindows() #destroy all windows
#cv2.destroyWindow('image') #destroy specific window
```

multiple images in one frame

```python
import numpy as np

img = cv2.imread('bottle.png')
resize_image = cv2.resize(img, (300, 300))
horizontal = np.hstack((resize_image, resize_image, resize_image))
vertical = np.vstack((horizontal, horizontal))
cv2.imshow('image', vertical)
cv2.waitKey(10000)
cv2.destroyAllWindows() #destroy all windows
```

load all images in directory 

```python
import os

images_list = os.listdir('images')
images_list
```

```python
for names in images_list:
    path = 'images'
    image_name = os.path.join(path, names)
    img = cv2.imread(image_name)
    cv2.imshow('image', img)
    cv2.waitKey(1000)
    cv2.destroyAllWindows() #destroy all windows
```
